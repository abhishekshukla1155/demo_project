**BREAST CANCER CLASSIFICATION WITH A SIMPLE NEURAL NETWORK**

Importing the Dependencies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn.datasets
from sklearn.model_selection import train_test_split
```

Data Collection and Processing

```python
# Loading the data form sklearn
breast_cancer_dataset = sklearn.datasets.load_breast_cancer()
```

```python
print(breast_cancer_dataset)
```

```python
# Loading the data to a data frame
data_frame = pd.DataFrame(breast_cancer_dataset.data, columns=breast_cancer_dataset.feature_names)
```

```python
# Print the first 5 rows of the dataframe
data_frame.head()
```

```python
# Adding the 'target' column to the data frome
data_frame['label'] = breast_cancer_dataset.target
```

```python
# Print last 5 rows of the dataframe
data_frame.tail()
```

```python
# Number of rows and columns in the dataset
data_frame.shape
```

```python
# Getting some information about the data
data_frame.info()
```

```python
# Statistical measures about the data
data_frame.describe()
```

```python
# Checking the distribution of target variable
data_frame['label'].value_counts()
```

1 -> Benign

0 -> Malignant

```python
data_frame.groupby('label').mean()
```

Separating the features and target

```python
x = data_frame.drop(columns='label', axis=1)
y = data_frame['label']
```

```python
print(x)
```

```python
print(y)
```

Splitting the data into Training data and Testing data

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=2)
```

```python
print(x.shape, x_train.shape, x_test.shape)
```

Standardize the data

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
x_train_std = scaler.fit_transform(x_train)
x_test_std = scaler.transform(x_test)
```

```python
print(x_train_std)
```

**BUILDING THE NEURAL NETWORK**

```python
# Importing Tensorflow and Keras
import tensorflow as tf
tf.random.set_seed(3)
from tensorflow import keras
```

```python
# Setting up the layers of Neural Network
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(30,)),
    keras.layers.Dense(20, activation='relu'),
    keras.layers.Dense(2, activation='sigmoid')
])
```

```python
# Compiling the Neural Network
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

Dog - 0

Cat - 1

Horse - 2

Dog - [1 0 0]

Cat - [0 1 0]

Horse - [0 0 1]

```python
# Training the Meural Network
history = model.fit(x_train, y_train, validation_split=0.1, epochs=10)
```

Visualizing accuracy and loss

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])

plt.title('model accuracy')
plt.ylabel('accuracy')
plt.xlabel('epoch')

plt.legend(['training data', 'validation data'], loc='lower right')
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])

plt.title('model loss')
plt.ylabel('loss')
plt.xlabel('epoch')

plt.legend(['training data', 'validation data'], loc='upper right')
```

Accuracy of the model on test data

```python
loss, accuracy = model.evaluate(x_test, y_test)
print(accuracy)
```

**Building the predictive System**

```python
input_data = (13.71,	20.83,	90.2,	577.9,	0.1189,	0.1645,	0.09366,	0.05985,	0.2196,	0.07451,	0.5835,	1.377,	3.856,	50.96,	0.008805,	0.03029,	0.02488,	0.01448,	0.01486,	0.005412,	17.06,	28.14,	110.6,	897,	0.1654,	0.3682,	0.2678,	0.1556,	0.3196,	0.1151)
# Change the input_data to a numpy array
input_data_as_numpy_array = np.asarray(input_data)

# Reshape the numpy array as we are predicting for one data point
input_data_reshaped = input_data_as_numpy_array.reshape(1, -1)

# Standardizing the input data
input_data_std = scaler.transform(input_data_reshaped)

prediction = model.predict(input_data_std)
print(prediction)

prediction_labels = [np.argmax(prediction)]

print(prediction_labels)

if(prediction_labels[0] == 0):
  print('The tumor is Malignant')

else:
  print('The tumor is Benign')
```
